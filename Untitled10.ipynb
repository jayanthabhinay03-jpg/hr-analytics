```python
plt.figure(figsize=(8, 6))
sns.histplot(df['Age'], bins=30, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x='YearsAtCompany', y='MonthlyIncome', data=df)
plt.title('Monthly Income vs. Years at Company')
plt.xlabel('Years at Company')
plt.ylabel('Monthly Income')
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(x='YearsAtCompany', y='MonthlyIncome', hue='Attrition', data=df, s=100)
plt.title('Monthly Income vs. Years at Company by Attrition')
plt.xlabel('Years at Company')
plt.ylabel('Monthly Income')
plt.show()
```

```python
plt.figure(figsize=(15, 10))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', linewidths=.5)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```
